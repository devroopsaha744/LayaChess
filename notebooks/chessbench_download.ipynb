```python
# Downloads ChessBench action-value files into /kaggle/working -> turn the output into a Kaggle dataset.
# Settings: Internet ON, accelerator None (CPU is enough).
import os, subprocess
N_TRAIN_SHARDS = 2          # each ~0.5 GB; total output must stay under ~19 GB
BASE = "https://storage.googleapis.com/searchless_chess/data"
OUT = "/kaggle/working"
files = {"test_action_value.bag": f"{BASE}/test/action_value_data.bag"}
for i in range(N_TRAIN_SHARDS):
    files[f"train_action_value_{i:05d}.bag"] = f"{BASE}/train/action_value-{i:05d}-of-02148_data.bag"
for name, url in files.items():
    subprocess.run(["wget", "-q", "-O", f"{OUT}/{name}", url], check=True)
    print(name, round(os.path.getsize(f"{OUT}/{name}") / 1e6), "MB")
```
